```python
from sklearn.datasets import load_iris

iris = load_iris()

print("Number of instances (rows):", iris.data.shape[0])
print("Number of features (Columns):", iris.data.shape[1])
print("Feature Names:", iris.feature_names)
print("Target names:", iris.target_names)
```

```python
df = pd.DataFrame(iris.data, columns =  iris.feature_names)

df['target']=iris.target
df['species']= df['target'].apply(lambda x: iris.target_names[x])

df.head()
```

1. Data Types Analysis

a.

1. A customer database stored in MySQL: Structured data. Highly organized in strict tables with predefined rows, columns, and relations.

2. JSON data returned from a REST API: Semi-structured data. Does not use tabular schemas but contains key-value markers and tags to separate data elements.

3. CCTV surveillance video: Unstructured data. Completely unorganized raw binary files without a predefined conceptual model or schema.

4. Emails containing headers and message bodies: Semi-structured data. Combines structured metadata headers (To, From, Date) with unstructured conversational text bodies.

5. An Excel spreadsheet containing student marks: Structured data. Organized in a clear grid format consisting of fixed rows and labeled columns.

Part D – Understanding the concepts
1. Why is data understanding essential before training a machine learning model?
- Data understanding helps identify missing values, outliers, duplicates, feature distributions, and relationships between variables. It ensures the data is clean and suitable for building an accurate and reliable machine learning model.

2. What is the difference between univariate and bivariate analysis?
- Univariate Analysis
Examines a single variable at a time.
Examples: histogram, boxplot, mean, median, variance.

- Bivariate Analysis
Examines the relationship between two variables.
Examples: scatter plot, correlation analysis, grouped boxplots.

3. Why should duplicate records be removed?
- Duplicate records can bias the model by giving repeated observations more influence, leading to inaccurate analysis and reduced model generalization.

4. Why can missing values reduce model performance?
- Missing values reduce model performance because they cause incomplete data and can lead to incorrect predictions. Filling missing values properly helps improve model accuracy.

5. What is data leakage? Give one example.
- Data leakage happens when information from the test data is accidentally used during training, giving unrealistically high performance.
Example: Scaling the entire dataset before splitting into training and testing data.

6. Why must train-test splitting be performed before preprocessing?
- Train-test splitting should be done before preprocessing to prevent data leakage. This ensures that preprocessing is learned only from the training data and the model is tested fairly.
